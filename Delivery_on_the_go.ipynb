```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Create a function to group ratings into clear buckets
def bucket_ratings(rating):
    if rating < 3.0:
        return 'Low (< 3.0)'
    elif rating < 4.0:
        return 'Medium (3.0 - 3.9)'
    elif rating < 4.5:
        return 'Good (4.0 - 4.4)'
    else:
        return 'Excellent (4.5 - 5.0)'

# 2. Apply the buckets to a new column
df['Rating_Group'] = df['Delivery_person_Ratings'].apply(bucket_ratings)

# 3. Plot the new grouped data
plt.figure(figsize=(8, 5))
order = ['Low (< 3.0)', 'Medium (3.0 - 3.9)', 'Good (4.0 - 4.4)', 'Excellent (4.5 - 5.0)']
sns.barplot(x='Rating_Group', y='Time_taken(min)', data=df, order=order, palette='muted')

plt.title('Average Delivery Time by Rating Group')
plt.ylabel('Avg Time taken (min)')
plt.xlabel('Rider Rating Group')
plt.show()
```

```python
import numpy as np

def haversine_vectorized(lat1, lon1, lat2, lon2):
    R = 6371.0  # Earth radius in kilometers

    # Convert all columns from degrees to radians
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = np.sin(dlat / 2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon / 2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1 - a))

    return R * c

```

```python
# This will now work perfectly with your columns:
df['Distance'] = haversine_vectorized(
    df['Restaurant_latitude'],
    df['Restaurant_longitude'],
    df['Delivery_location_latitude'],
    df['Delivery_location_longitude']
)
```

```python
def daytime(hour):
    if hour >=0 and hour <=12 :
        return 'Morning'
    elif hour >=13 and hour <=15 :
        return 'Afternoon'
    elif hour >=15 and hour <=18 :
        return 'Evening'
    else :
        return 'Night'
```
